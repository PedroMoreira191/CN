# Gauss–Seidel — Cálculo Numérico
Sistema:

10x − y + 2z = 6

−x + 11y − z = 22

2x − y + 10z = −10

## Questão 5 — Implementação

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[10, -1, 2],
              [-1, 11, -1],
              [2, -1, 10]], dtype=float)
b = np.array([6, 22, -10], dtype=float)

def gauss_seidel(A, b, x0, tol=1e-5, max_iter=1000):
    n = len(b)
    x = x0.copy()
    historico = [(0, *x, np.nan)]
    for k in range(1, max_iter + 1):
        x_ant = x.copy()
        for i in range(n):
            soma = sum(A[i, j] * x[j] for j in range(n) if j != i)
            x[i] = (b[i] - soma) / A[i, i]   # usa valores já atualizados
        erro = np.max(np.abs(x - x_ant))
        historico.append((k, *x, erro))
        if erro < tol:
            break
    return x, k, historico

x, k, hist = gauss_seidel(A, b, np.zeros(3))
print(f"Número de iterações: {k}")
print(f"Solução aproximada: x = {x[0]:.8f}, y = {x[1]:.8f}, z = {x[2]:.8f}")
```

## Questão 6 — Tabela

```python
print(f"{'k':>2} {'x(k)':>12} {'y(k)':>12} {'z(k)':>12} {'||x(k)-x(k-1)||inf':>20}")
for kk, xx, yy, zz, e in hist:
    print(f"{kk:>2} {xx:>12.8f} {yy:>12.8f} {zz:>12.8f} {e:>20.3e}")
```

## Questão 7a — Evolução de x, y e z

```python
H = np.array(hist)
plt.figure(figsize=(7, 4))
plt.plot(H[:, 0], H[:, 1], 'o-', label='x(k)')
plt.plot(H[:, 0], H[:, 2], 's-', label='y(k)')
plt.plot(H[:, 0], H[:, 3], '^-', label='z(k)')
plt.xlabel('Iteração k'); plt.ylabel('Valor')
plt.title('Evolução das aproximações'); plt.grid(True); plt.legend()
plt.tight_layout(); plt.show()
```

## Questão 7b — Erro entre iterações (escala log)

```python
plt.figure(figsize=(7, 4))
plt.semilogy(H[1:, 0], H[1:, 4], 'o-', color='red')
plt.xlabel('Iteração k'); plt.ylabel(r'$\|x^{(k)}-x^{(k-1)}\|_\infty$')
plt.title('Erro entre iterações (escala log)'); plt.grid(True, which='both')
plt.tight_layout(); plt.show()
```
